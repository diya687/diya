```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy  as np
cars_df=pd.read_csv('/content/mtcars (2).csv')
print(cars_df)
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
print(a)
plt.xlabel('mpg')
plt.ylabel('weight')
plt.title('relationhip between the weight of the car and mpg')
plt.show()
trans=cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel('transmission')
plt.title('barplottocheckthefrequencydistributionoftransmissiontype')
plt.show()
cars_df['mpg'].plot(kind='box')
plt.xlabel('mpg')
plt.title('boxandwhiskerplotofmpg')
plt.show()
minimum_value=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maximum_value=np.max(cars_df['mpg'])
print('minimum_value;',minimum_value)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maximum_value:',maximum_value)

```
